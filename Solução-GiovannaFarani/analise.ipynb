```python
funcionarios.duplicated().sum()
```

```python
funcionarios["data_admissao"] = funcionarios["data_admissao"].str.replace("/","-")
```

```python
funcionarios["data_admissao"] = pd.to_datetime(
    funcionarios["data_admissao"],
    format="mixed",
)
```

```python
funcionarios["data_promocao"] = funcionarios["data_promocao"].str.replace("/","-")

funcionarios["data_promocao"] = pd.to_datetime(
    funcionarios["data_promocao"],
    format="mixed",
)
```

```python
funcionarios_filtrados = funcionarios[
    (funcionarios["genero"] != "Feminino") &
    (funcionarios["genero"] != "Masculino") &
    (funcionarios["genero"] != "Não-binário")
]
funcionarios_filtrados
```

```python
funcionarios["genero"] = funcionarios["genero"].replace("Fem","Feminino")
```

```python
funcionarios["status_atual"].unique()
```

```python
funcionarios["data_promocao"].dropna()
```

```python
funcionarios["salario_base"] = funcionarios["salario_base"].str.replace(
    r"^R\$\s*",
    "",
    regex=True
)
funcionarios
```

```python
funcionarios["salario_base"] = funcionarios["salario_base"].str.replace(".",",")
funcionarios
```
